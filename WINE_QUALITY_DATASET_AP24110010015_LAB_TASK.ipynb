```python
from google.colab import files

uploaded = files.upload()
import pandas as pd
import numpy as np

# --------------------------------------------------
# 1. Load dataset
import pandas as pd

df = pd.read_csv("WineQT.csv")

print("Original dataset shape:", df.shape)
print(df.head())

# --------------------------------------------------
# 2. Remove Id column
# --------------------------------------------------

X = df.drop(columns=["Id"])

print("\nFeatures used for PCA:")
print(X.columns.tolist())

print("\nShape after removing Id:", X.shape)


# --------------------------------------------------
# 3. Convert DataFrame to NumPy array
# --------------------------------------------------

X = X.to_numpy(dtype=float)


# --------------------------------------------------
# 4. Standardize the data
# --------------------------------------------------

mean = np.mean(X, axis=0)
std = np.std(X, axis=0)

Z = (X - mean) / std

print("\nStandardized data:")
print(Z[:5])


# --------------------------------------------------
# 5. Calculate covariance matrix
# --------------------------------------------------

covariance_matrix = np.cov(X, rowvar=False)

print("\nCovariance Matrix:")
print(covariance_matrix)

print("\nCovariance Matrix Shape:")
print(covariance_matrix.shape)


# --------------------------------------------------
# 6. Calculate eigenvalues and eigenvectors
# --------------------------------------------------

eigenvalues, eigenvectors = np.linalg.eigh(covariance_matrix)


# --------------------------------------------------
# 7. Sort eigenvalues in descending order
# --------------------------------------------------

indices = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[indices]
eigenvectors = eigenvectors[:, indices]


print("\nEigenvalues:")
print(eigenvalues)


# --------------------------------------------------
# 8. Select first two principal components
# --------------------------------------------------

PC1 = eigenvectors[:, 2]
PC2 = eigenvectors[:, 3]

print("\nPC1:")
print(PC1)

print("\nPC2:")
print(PC2)


# --------------------------------------------------
# 9. Create projection matrix
# --------------------------------------------------

W = eigenvectors[:, :2]

print("\nProjection Matrix:")
print(W)


# --------------------------------------------------
# 10. Transform original data into 2 dimensions
# --------------------------------------------------

X_pca = X @ W


# --------------------------------------------------
# 11. Create final 2D DataFrame
# --------------------------------------------------

pca_df = pd.DataFrame(
    X_pca,
    columns=["PC1", "PC2"]
)

print("\nReduced Dataset:")
print(pca_df.head())

print("\nReduced Dataset Shape:")
print(pca_df.shape)


# --------------------------------------------------
# 12. Explained variance
# --------------------------------------------------

explained_variance_ratio = (
    eigenvalues / np.sum(eigenvalues)
)

print("\nExplained Variance Ratio:")
print(explained_variance_ratio)

print("\nPC1 explained variance:",
      explained_variance_ratio[0] * 100, "%")

print("PC2 explained variance:",
      explained_variance_ratio[1] * 100, "%")

print("Total explained variance:",
      np.sum(explained_variance_ratio[:2]) * 100, "%")


# --------------------------------------------------
# 13. Save reduced dataset
# --------------------------------------------------

pca_df.to_csv("WineQT_PCA_2D.csv", index=False)

print("\nReduced dataset saved as WineQT_PCA_2D.csv")
```
