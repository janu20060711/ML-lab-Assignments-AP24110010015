9.Using the `marks` array above:
1. Extract all marks that are failing (below 40).
2. Count how many students scored above 80 (hint: `.sum()` on a boolean array counts `True` values).
3. Replace all failing marks with 0 using `np.where`.


```python
import numpy as np

# Create the marks array
marks = np.array([55, 62, 48, 91, 73, 35, 28, 85])

# 1. Extract all marks that are failing (below 40)
failing_marks = marks[marks < 40]
print("Failing marks:", failing_marks)

# 2. Count how many students scored above 80
# The condition creates True/False values.
# .sum() counts the True values.
above_80 = (marks > 80).sum()
print("Students who scored above 80:", above_80)

# 3. Replace all failing marks with 0 using np.where()
# If the mark is below 40, replace it with 0.
# Otherwise, keep the original mark.
updated_marks = np.where(marks < 40, 0, marks)
print("Marks after replacing failing marks with 0:", updated_marks)
```

15.Calculate the average brightness of your image using `np.mean()`. Then calculate the average brightness of just the red channel, just the green channel, and just the blue channel. Which channel is brightest? Write your answer below the code.

```python
import numpy as np
import matplotlib.image as mpimg

# Load the image
img = mpimg.imread("download.jpg")

# Calculate the average brightness of the entire image
# np.mean() calculates the average of all pixel values
average_brightness = np.mean(img)

# Extract the Red, Green, and Blue channels
red_channel = img[:, :, 0]
green_channel = img[:, :, 1]
blue_channel = img[:, :, 2]

# Calculate the average brightness of each colour channel
red_brightness = np.mean(red_channel)
green_brightness = np.mean(green_channel)
blue_brightness = np.mean(blue_channel)

# Display the results
print("Average brightness of the image:", average_brightness)
print("Average brightness of Red channel:", red_brightness)
print("Average brightness of Green channel:", green_brightness)
print("Average brightness of Blue channel:", blue_brightness)

# Find which channel has the highest average brightness
brightness = {
    "Red": red_brightness,
    "Green": green_brightness,
    "Blue": blue_brightness
}

brightest_channel = max(brightness, key=brightness.get)

print("Brightest channel:", brightest_channel)
```

16.Convert the image into pure black-and-white using `np.where()`.

### Student Practice
1. Convert the image to grayscale by averaging the 3 channels (`axis=2`).
2. Use `np.where(gray > 128, 255, 0)` to create a binary (black/white) version.
3. Display the result.
4. Try a different threshold (e.g. 100 or 180) and observe how it changes the result.


```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Load the image
img = mpimg.imread("download.jpg")

# Convert the image to grayscale
# axis=2 means we calculate the average of Red, Green, and Blue channels
gray = np.mean(img[:, :, :3], axis=2)

# Convert grayscale image into pure black and white
# If brightness is greater than 128 -> white (255)
# Otherwise -> black (0)
binary = np.where(gray > 128, 255, 0)

# Display the black-and-white image
plt.imshow(binary, cmap="gray")
plt.axis("off")
plt.title("Binary Black-and-White Image")
plt.show()

# Try a different threshold value
# Here, pixels brighter than 180 become white
binary_180 = np.where(gray > 180, 255, 0)

# Display the result with threshold 180
plt.imshow(binary_180, cmap="gray")
plt.axis("off")
plt.title("Binary Image - Threshold 180")
plt.show()
```

21.Pick two sentences of your own (write them below) — one pair that should be *similar* in meaning/topic, and one pair that should be *very different*. Compute the cosine similarity for both pairs and confirm your intuition matches the numbers.

```python
# Import the required libraries
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


# ---------------------------------------------------------
# Pair 1: Sentences with similar meaning/topic
# ---------------------------------------------------------

sentence1 = "I love playing football."
sentence2 = "I enjoy playing soccer."


# Put both sentences into a list
sentences_similar = [sentence1, sentence2]

# Convert sentences into TF-IDF vectors
vectorizer = TfidfVectorizer()
vectors = vectorizer.fit_transform(sentences_similar)

# Calculate cosine similarity
similarity1 = cosine_similarity(vectors[0], vectors[1])[0][0]

print("Pair 1 - Similar Sentences")
print("Sentence 1:", sentence1)
print("Sentence 2:", sentence2)
print("Cosine Similarity:", similarity1)


# ---------------------------------------------------------
# Pair 2: Sentences with very different meaning/topic
# ---------------------------------------------------------

sentence3 = "I love playing football."
sentence4 = "The sun rises in the east."


# Put both sentences into a list
sentences_different = [sentence3, sentence4]

# Convert sentences into TF-IDF vectors
vectors2 = vectorizer.fit_transform(sentences_different)

# Calculate cosine similarity
similarity2 = cosine_similarity(vectors2[0], vectors2[1])[0][0]

print("\nPair 2 - Very Different Sentences")
print("Sentence 1:", sentence3)
print("Sentence 2:", sentence4)
print("Cosine Similarity:", similarity2)


# ---------------------------------------------------------
# Confirm our intuition
# ---------------------------------------------------------

print("\nConclusion:")

if similarity1 > similarity2:
    print("The result matches our intuition.")
    print("The similar sentences have a higher cosine similarity.")
else:
    print("The result does not match our intuition.")
```

22.Take any paragraph of your choice (3-4 sentences, write it below as a string). Then:
1. Tokenize it.
2. Build its vocabulary and word-count vector (reuse `bag_of_words_vector`).
3. Use NumPy to find and print the **top 3 most frequent words** (hint: look at sorting — `np.argsort()` sorts indices by value, and slicing with `[::-1]` reverses order)

```python
import numpy as np

# Write a paragraph containing 3-4 sentences
paragraph = """
Python is a popular programming language. Python is easy to learn and useful.
Programming helps students solve problems. Learning programming improves logical thinking.
"""

# 1. Tokenize the paragraph into individual words
# Convert to lowercase so that "Python" and "python" are treated as the same word
tokens = paragraph.lower().split()

print("Tokens:")
print(tokens)

# 2. Build the vocabulary
# sorted(set()) keeps only unique words and sorts them
vocabulary = sorted(set(tokens))

print("\nVocabulary:")
print(vocabulary)

# Define the bag_of_words_vector function
# It counts how many times each vocabulary word appears
def bag_of_words_vector(tokens, vocabulary):
    return np.array([tokens.count(word) for word in vocabulary])

# Create the word-count vector
word_counts = bag_of_words_vector(tokens, vocabulary)

print("\nWord-count vector:")
print(word_counts)

# 3. Find the indices of the words sorted by their frequency
# np.argsort() gives indices from smallest to largest count
# [::-1] reverses the order, giving largest to smallest
sorted_indices = np.argsort(word_counts)[::-1]

# Select the top 3 most frequent words
top_3_indices = sorted_indices[:3]

print("\nTop 3 most frequent words:")

# Print each word along with its frequency
for index in top_3_indices:
    print(vocabulary[index], ":", word_counts[index])
```
