7.Column addition

```python
import pandas as pd

# Create a dictionary containing two Pandas Series
info = {
    'one': pd.Series(
        [1, 2, 3, 4, 5],
        index=['a', 'b', 'c', 'd', 'e']
    ),

    'two': pd.Series(
        [1, 2, 3, 4, 5, 6],
        index=['a', 'b', 'c', 'd', 'e', 'f']
    )
}

# Create a DataFrame using the dictionary
df = pd.DataFrame(info)

# Print the original DataFrame
print(df)

# Add a new column by passing a Pandas Series
print("Add new column by passing series")

# Create a new column 'three' with values 20, 40, 60
df['three'] = pd.Series(
    [20, 40, 60],
    index=['a', 'b', 'c']
)

# Print the DataFrame after adding the 'three' column
print(df)

# Add a new column using existing DataFrame columns
print("Add new column using existing DataFrame columns")

# Create 'four' by adding the values of 'one' and 'three'
df['four'] = df['one'] + df['three']

# Print the final DataFrame
print(df)
```

8.Column deletion

```python
import pandas as pd

# Create a dictionary containing two Pandas Series
info = {
    'one': pd.Series([1, 2], index=['a', 'b']),
    'two': pd.Series([1, 2, 3], index=['a', 'b', 'c'])
}

# Create a DataFrame using the dictionary
df = pd.DataFrame(info)

# Print the original DataFrame
print("The DataFrame:")
print(df)

# Delete the first column using the del function
print("Delete the first column:")
del df['one']

# Print the DataFrame after deleting column 'one'
print(df)

# Delete the remaining column using the pop() function
print("Delete another column:")
df.pop('two')

# Print the final DataFrame
print(df)
```

9.Selection by label

```python
import pandas as pd

# Create a dictionary containing two Pandas Series
# Each Series has values with corresponding index labels
info = {
    'one': pd.Series([1, 2, 3, 4, 5],
                     index=['a', 'b', 'c', 'd', 'e']),

    'two': pd.Series([1, 2, 3, 4, 5, 6],
                     index=['a', 'b', 'c', 'd', 'e', 'f'])
}

# Create a DataFrame using the dictionary
df = pd.DataFrame(info)

# Display the complete DataFrame
print("Data frame:")
print(df)

# Display all column values at index 'b'
print("b index elements are:")
print(df.loc['b'])

# Display all column values at index 'e'
print("e index elements are:")
print(df.loc['e'])
```

10.passing the integer location to an iloc function.

```python
# Import the pandas library
import pandas as pd

# Create a dictionary containing two Pandas Series
info = {
    # First column named 'one' with index labels
    'one': pd.Series([1, 2, 3, 4, 5],
                     index=['a', 'b', 'c', 'd', 'e']),

    # Second column named 'two' with index labels
    'two': pd.Series([1, 2, 3, 4, 5, 6],
                     index=['a', 'b', 'c', 'd', 'e', 'f'])
}

# Create a DataFrame using the dictionary
df = pd.DataFrame(info)

# Print the complete DataFrame
print("Data Frame:")
print(df)

# Print the row at integer position 3
# iloc[3] means the fourth row because indexing starts from 0
print("\nRow at position 3:")
print(df.iloc[3])
```

13.Deletion of rows

```python
# Import the pandas library
import pandas as pd

# Create the first DataFrame with two rows and columns x and y
a_info = pd.DataFrame([[4, 5], [6, 7]], columns=['x', 'y'])

# Create the second DataFrame with two rows and columns x and y
b_info = pd.DataFrame([[8, 9], [10, 11]], columns=['x', 'y'])

# Concatenate (combine) both DataFrames vertically
a_info = pd.concat([a_info, b_info])

# Print the combined DataFrame
print("Combined DataFrame:")
print(a_info)

# Drop all rows with index label 0
a_info = a_info.drop(0)

# Print the DataFrame after dropping the row with label 0
print("Drop the row with label 0:")
print(a_info)
```

16.dataframe.duplicates

17.Null values identification in given dataset:

```python
# Import the pandas library
import pandas as pd

# Import the NumPy library
import numpy as np

# Create a DataFrame with Person names and Age values
# np.nan represents a missing or null value
info = pd.DataFrame({
    "Person": ["Siva", "Pavan", "Gopal", "Raghu"],
    "Age": [27., 29, np.nan, 32]
})

# Print the original DataFrame
print(info)

# Count the number of non-null values in each column
print(info.count())

# Print a message indicating null values
print("Null values")

# Check each value in the DataFrame
# Returns True if the value is null, otherwise False
print(info.isnull())

# Count the total number of null values in each column
print(info.isnull().sum())
```

18.Sorting order

```python
# Import the pandas library
import pandas as pd

# Import the NumPy library
import numpy as np

# Create a DataFrame with Person names and Age values
# np.nan represents a missing or null value
info = pd.DataFrame({
    "Person": ["Siva", "Pavan", "Gopal", "Raghu"],
    "Age": [27., 29, np.nan, 32]
})

# Print the original DataFrame
print(info)

# Print a message for sorting in ascending index order
print("Sorting Order")

# Sort the DataFrame based on the index in ascending order
print(info.sort_index())

# Print a message for sorting in descending index order
print("Descending Order")

# Sort the DataFrame based on the index in descending order
print(info.sort_index(ascending=False))
```

19.Aggreagate functions

```python
import pandas as pd

# Read Iris dataset using the online link
df = pd.read_csv("https://raw.githubusercontent.com/fslaborg/datasets/main/data/iris.csv")

# Print the complete dataset
print(df)

# Calculate mean values for each species
mean_df = df.groupby('species').mean()

print("Mean values for each species:")
print(mean_df)

# Calculate sum of values for each species
sum_df = df.groupby('species').sum()

print("\nSum of values for each species:")
print(sum_df)
```
